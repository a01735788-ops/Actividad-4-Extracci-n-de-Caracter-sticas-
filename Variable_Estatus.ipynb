```python

import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 1. Cargamos el archivo Base de Leads
Base_Leads = pd.read_excel('Base_de_Leads.xlsx')
```

```python
# 2. la columna de estatus viene sin nombre, así que la renombramos a 'Estatus'
Base_Leads = Base_Leads.rename(columns={' ': 'Estatus'})
```

```python
# 3. Obtenemos un análisis univariado de la variable categórica 'Estatus'
#    value_counts() cuenta cuántas veces aparece cada estatus en la base
Tabla_freq = Base_Leads['Estatus'].value_counts().reset_index()
Tabla_freq
```

```python
# 4. Filtramos los valores más relevantes (igual que en clase)
#    Nos quedamos solo con categorías que aparecen más de 1 vez
Filtro = Tabla_freq[Tabla_freq['count'] > 1]
Filtro
```

```python
# 5. Ajustamos el índice de mi dataframe
Filtro_index = Filtro.set_index('Estatus')
Filtro_index
```

```python
# 6. gráfico de BARRAS
Filtro_index.plot(kind='bar', width=0.8, figsize=(10, 4), color='red', rot=45)
plt.title('Frecuencia de Estatus')
plt.xlabel('Estatus')
plt.ylabel('Frecuencia')
plt.tight_layout()
plt.show()
```
