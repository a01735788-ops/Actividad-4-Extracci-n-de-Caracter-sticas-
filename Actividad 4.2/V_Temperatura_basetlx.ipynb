```python
# Importamos las librerías necesarias
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Cargamos el archivo Base de Tlaxcala
Base_Tlaxcala = pd.read_excel('Base_de_Tlaxcala.xlsx')
Base_Tlaxcala.head()
```

```python
# Obtenemos un análisis univariado de la variable categórica 'Temperatura'
Tabla_freq = Base_Tlaxcala['Temperatura'].value_counts().reset_index()
Tabla_freq
```

```python
# Filtramos los valores más relevantes (nos quedamos con categorías con más de 1 registro)
Filtro = Tabla_freq[Tabla_freq['count'] > 1]
Filtro
```

```python
# Ajustamos el índice de mi dataframe
Filtro_index = Filtro.set_index('Temperatura')
Filtro_index
```

```python
# Primero: definimos la función de etiqueta (si no la has corrido ya)
total = Filtro_index['count'].sum()

def etiqueta(pct):
    cantidad = int(round(pct / 100 * total))
    return f'{pct:.1f}%\n({cantidad})'

# Separamos un poco las rebanadas chiquitas para que no se encimen sus etiquetas
explode = [0.08 if valor <= 2 else 0 for valor in Filtro_index['count']]

Filtro_index['count'].plot(
    kind='pie',
    figsize=(8, 6),
    shadow=True,
    autopct=etiqueta,
    explode=explode,
    pctdistance=0.75,
    labeldistance=1.1,
    colors=['#4C72B0', '#DD8452', '#55A868']
)
plt.title('Distribución de Temperatura')
plt.ylabel('')
plt.show()
```
