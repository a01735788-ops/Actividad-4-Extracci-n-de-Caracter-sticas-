```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 1. Cargamos el archivo Base de Leads
Base_Leads = pd.read_excel('Base_de_Leads.xlsx')
```

```python
# 2. Obtenemos un análisis univariado de la variable categórica 'Temperatura'
#    value_counts() cuenta cuántas veces aparece cada nivel de interés
Tabla_freq = Base_Leads['Temperatura'].value_counts().reset_index()
Tabla_freq
```

```python
# 3. Ajustamos el índice de mi dataframe (igual que en el resto del notebook)
Filtro_index = Tabla_freq.set_index('Temperatura')
Filtro_index
```

```python
# 4. ¿Por qué un gráfico de PASTEL (pie)?
#    'Temperatura' solo tiene 3 categorías, y lo que más interesa comunicar
#    aquí es la PROPORCIÓN (el %) que representa cada nivel de intención de
#    compra dentro del total de leads (ej. qué % está "Muy Interesado").
#    Con pocas categorías, el pastel se lee muy fácil y resalta de inmediato
#    que casi el 90% de los leads están en "Interés Medio".
Filtro_index.plot(kind='pie', y='count', autopct='%1.1f%%', figsize=(6, 6),
                   startangle=90, legend=False)
plt.title('Distribución de Temperatura')
plt.ylabel('')  # quitamos la etiqueta del eje y, no aporta nada en un pastel
plt.show()
```
