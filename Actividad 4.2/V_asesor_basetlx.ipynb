```python
# Importamos las librerías necesarias
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Cargamos el archivo Base de Tlaxcala
Base_Tlaxcala = pd.read_excel('Base_de_Tlaxcala.xlsx')
Base_Tlaxcala.head()
```

```python
# Obtenemos un análisis univariado de la variable categórica 'Asesor'
Tabla_freq = Base_Tlaxcala['Asesor'].value_counts().reset_index()
Tabla_freq
```

```python
# Filtramos los valores más relevantes (nos quedamos con categorías con más de 1 registro)
Filtro = Tabla_freq[Tabla_freq['count'] > 1]
Filtro
```

```python
# Ajustamos el índice de mi dataframe
Filtro_index = Filtro.set_index('Asesor')
Filtro_index
```

```python
# 6. Antes de graficar, renombramos la columna para que la leyenda sea más clara
Filtro_index = Filtro_index.rename(columns={'count': 'Número de leads'})

ax = Filtro_index.plot(kind='bar', width=0.8, figsize=(10, 4), color='blue', rot=45)
plt.title('Número de leads atendidos por asesor')
plt.xlabel('Asesor')
plt.ylabel('Número de leads')
ax.bar_label(ax.containers[0])

# Un poco de espacio extra arriba para que la etiqueta "31" no quede pegada al borde
ax.set_ylim(0, Filtro_index['Número de leads'].max() * 1.15)

plt.tight_layout()
plt.show()
```
