```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Cargamos el archivo Base de Tlaxcala
Base_Tlaxcala = pd.read_excel('Base_de_Tlaxcala.xlsx')
Base_Tlaxcala.head()
```

```python
# Obtenemos un análisis univariado de la variable categórica 'Producto'
Tabla_freq = Base_Tlaxcala['Producto'].value_counts().reset_index()
Tabla_freq
```

```python
# Filtramos los valores más relevantes (nos quedamos con categorías con más de 1 registro)
Filtro = Tabla_freq[Tabla_freq['count'] > 1]
Filtro
```

```python
# Ajustamos el índice de mi dataframe
Filtro_index = Filtro.set_index('Producto')
Filtro_index
```

```python
# Gráfico de PASTEL
total = Filtro_index['count'].sum()

def etiqueta(pct):
    cantidad = int(round(pct / 100 * total))
    return f'{pct:.1f}%\n({cantidad})'

Filtro_index['count'].plot(kind='pie', figsize=(8, 6), shadow=True, autopct=etiqueta)
plt.title('Distribución de Producto')
plt.ylabel('')
plt.show()
```
