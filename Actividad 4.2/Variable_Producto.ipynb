```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 1. Cargamos el archivo Base de Leads
Base_Leads = pd.read_excel('Base_de_Leads.xlsx')
```

```python
# 2. Obtenemos un análisis univariado de la variable categórica 'Producto'
#    value_counts() cuenta cuántas veces aparece cada producto en la base
Tabla_freq = Base_Leads['Producto'].value_counts().reset_index()
Tabla_freq
```

```python
# 3. Ajustamos el índice de mi dataframe (igual que en el resto del notebook)
Filtro = Tabla_freq[Tabla_freq['count'] > 1]
Filtro
```

```python
Filtro_index = Filtro.set_index('Producto')
Filtro_index
```

```python
# Renombramos la columna antes de graficar, para que la leyenda sea más clara
Filtro_index = Filtro_index.rename(columns={'count': 'Número de leads'})

ax = Filtro_index.plot(kind='bar', width=0.8, figsize=(10, 4), color='red', rot=45)
plt.title('Número de leads por producto de interés')
plt.xlabel('Producto')
plt.ylabel('Número de leads')
ax.bar_label(ax.containers[0])

# Espacio extra arriba para que la etiqueta de la barra más alta no quede pegada al borde
ax.set_ylim(0, Filtro_index['Número de leads'].max() * 1.15)

plt.tight_layout()
plt.show()
```
