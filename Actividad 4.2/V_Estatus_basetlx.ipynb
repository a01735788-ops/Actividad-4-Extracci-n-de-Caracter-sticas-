```python
# Importamos las librerías necesarias
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Cargamos el archivo Base de Tlaxcala
Base_Tlaxcala = pd.read_excel('Base_de_Tlaxcala.xlsx')
Base_Tlaxcala.head()
```

```python
# Obtenemos un análisis univariado de la variable categórica 'Estatus'
Tabla_freq = Base_Tlaxcala['Estatus'].value_counts().reset_index()
Tabla_freq
```

```python
# Filtramos los valores más relevantes (nos quedamos con categorías con más de 1 registro)
Filtro = Tabla_freq[Tabla_freq['count'] > 1]
Filtro
```

```python
# Ajustamos el índice de mi dataframe
Filtro_index = Filtro.set_index('Estatus')
Filtro_index
```

```python
# Gráfico de BARRAS
ax = Filtro_index.plot(kind='bar', width=0.8, figsize=(10, 4), color='pink', rot=45)
plt.title('Frecuencia de Estatus')
plt.xlabel('Estatus')
plt.ylabel('Frecuencia')
ax.bar_label(ax.containers[0])
plt.tight_layout()
plt.show()
```
