```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 1. Cargamos el archivo Base de Leads
#    (si ya tienes 'Base_Leads' cargado en tu notebook, puedes omitir esta línea)
Base_Leads = pd.read_excel('Base_de_Leads.xlsx')
```

```python
# 2. Obtenemos un análisis univariado de la variable categórica 'Producto'
#    value_counts() cuenta cuántas veces aparece cada producto en la base
Tabla_freq = Base_Leads['Producto'].value_counts().reset_index()
Tabla_freq
```

```python
# 3. Ajustamos el índice de mi dataframe (igual que en el resto del notebook)
Filtro_index = Tabla_freq.set_index('Producto')
Filtro_index
```

```python
# 4. ¿Por qué un gráfico de BARRAS?
#    'Producto' tiene 7 categorías distintas (nominales, sin orden natural),
#    y sus frecuencias son muy distintas entre sí (de 107 hasta 1).
#    Un gráfico de barras es el más adecuado porque permite comparar
#    de forma clara y directa la altura de cada categoría, incluso cuando
#    hay muchas y con valores muy chicos (algo que un gráfico de pastel
#    no lograría mostrar bien, ya que las categorías pequeñas casi
#    no se alcanzarían a ver).
Filtro_index.plot(kind='bar', width=0.8, figsize=(10, 4), color='steelblue')
plt.title('Frecuencia de Producto')
plt.xlabel('Producto')
plt.ylabel('Frecuencia')
plt.xticks(rotation=45, ha='right')  # rotamos las etiquetas para que no se encimen
plt.tight_layout()
plt.show()
```
