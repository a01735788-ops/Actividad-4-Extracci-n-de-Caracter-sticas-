```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 1. Cargamos el archivo Base de Leads
Base_Leads = pd.read_excel('Base_de_Leads.xlsx')
```

```python
# 2. Obtenemos un análisis univariado de la variable categórica 'Temperatura'
#    value_counts() cuenta cuántas veces aparece cada nivel de interés
Tabla_freq = Base_Leads['Temperatura'].value_counts().reset_index()
Tabla_freq
```

```python
# 3. Ajustamos el índice de mi dataframe
Filtro = Tabla_freq[Tabla_freq['count'] > 1]
Filtro
```

```python
# 4. Ajustamos el índice de mi dataframe
Filtro_index = Filtro.set_index('Temperatura')
Filtro_index
```

```python
# 4. gráfico de PASTEL
Filtro_index['count'].plot(kind='pie', figsize=(8, 6), shadow=True, autopct='%0.1f %%')
plt.title('Distribución de Temperatura')
plt.ylabel('')  # quitamos la etiqueta del eje y, no aporta nada en un pastel
plt.show()
```
